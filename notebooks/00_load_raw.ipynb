```python
import sqlite3
import pandas as pd
import os

# Paths relative to the notebook folder
DB_PATH = "E:/Sidra/JupyterProjects/RiyalPay-Project/riyalpay.db"
BASE_DIR = "E:/Sidra/JupyterProjects/RiyalPay-Project/data"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
print(f"✅ Database created at {DB_PATH}\n")

files = {
    "raw_users": "users_raw.csv",
    "raw_experiment_assignments": "experiment_assignments_raw.csv",
    "raw_events_pre_migration": "events_pre_migration_raw.csv",
    "raw_events_post_migration": "events_post_migration_raw.csv",
    "raw_transactions": "transactions_raw.csv",
}

for table, fname in files.items():
    filepath = os.path.join(BASE_DIR, fname)
    df = pd.read_csv(filepath)
    df.to_sql(table, conn, if_exists="replace", index=False)
    print(f"Loaded {table}: {len(df):,} rows")

conn.close()
print("\nriyalpay.db successfuly created!")
```
